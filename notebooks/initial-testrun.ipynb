# Basic Usage Example

```python
import torch
from PIL import Image

from sam3.model.sam3_image_processor import Sam3Processor
from sam3.model_builder import build_sam3_image_model
```

```python
## Fixes problems with float32 / bfloat16 mismatch problems causing errors in the Image Path
torch.autocast("cuda", dtype=torch.bfloat16).__enter__()
```

```python
# Load the model
model = build_sam3_image_model()
processor = Sam3Processor(model)
```

## Image Path

```python
# Load an image
image = Image.open("../assets/images/test_image.jpg")

inference_state = processor.set_image(image)

# Prompt the model with text
output = processor.set_text_prompt(state=inference_state, prompt="girl")

# Get the masks, bounding boxes, and scores
masks, boxes, scores = output["masks"], output["boxes"], output["scores"]
```

```python
print("Boxes:\n")
print(boxes)

print("Scores:\n")
print(scores)

print("Masks:\n")
print(masks)
```

## Video Path

```python
from sam3.model_builder import build_sam3_video_predictor

video_predictor = build_sam3_video_predictor()
video_path = "../assets/videos/bedroom.mp4"  # a JPEG folder or an MP4 video file

# Start a session
response = video_predictor.handle_request(
    request={
        "type": "start_session",
        "resource_path": video_path,
    }
)

response = video_predictor.handle_request(
    request={
        "type": "add_prompt",
        "session_id": response["session_id"],
        "frame_index": 0,  # Arbitrary frame index
        "text": "person",
    }
)

output = response["outputs"]
```

```python
print("Output:\n")
print(output)
```
